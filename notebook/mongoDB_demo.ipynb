##### This is a notebook just to push data from my local to MONGODB so we can pull it later

```python
import os
from dotenv import load_dotenv
import pandas as pd
import pymongo
```

```python
#loading connection url from my local .env file so it remains private. 
load_dotenv()
MONGODB_URI=os.getenv('uri')
```

```python
df=pd.read_csv('data.csv')
data=df.to_dict(orient='records')
#data
```

```python
DB_NAME='Proj1'
Collection_name='Proj1-Data'
Connection_URL=MONGODB_URI
```

```python
client=pymongo.MongoClient(Connection_URL)
database=client[DB_NAME]
collection=database[Collection_name]
```
